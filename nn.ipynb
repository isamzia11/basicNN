```python
import numpy as np
```

```python

inputs = np.array([
    [73, 67, 43],
    [85, 72, 50],
    [60, 55, 40],
    [90, 80, 70],
    [45, 50, 35],
    [78, 65, 55],
    [88, 75, 60],
    [52, 48, 42],
    [95, 85, 75],
    [68, 62, 50],
    [77, 70, 58],
    [82, 74, 65],
    [55, 60, 45],
    [92, 88, 78],
    [70, 68, 52]
],dtype=float)

print(inputs)

print(inputs.shape)
```

```python
import torch
```

```python
input = torch.from_numpy(inputs)
```

```python
targets = np.array([
    [73, 67, 43],
    [85, 72, 50],
    [60, 55, 40],
    [90, 80, 70],
    [45, 50, 35],
    [78, 65, 55],
    [88, 75, 60],
    [52, 48, 42],
    [95, 85, 75],
    [68, 62, 50],
    [77, 70, 58],
    [82, 74, 65],
    [55, 60, 45],
    [92, 88, 78],
    [70, 68, 52]
], dtype=np.float32)

print(targets.shape)
print(targets.dtype)
```

```python
targets = torch.from_numpy(targets)
```

```python
import torch.nn as nn

### dataset, dataloader 
```

```python
from torch.utils.data import TensorDataset
```

```python
train_ds = TensorDataset(input, targets)
```

```python
train_ds[0:3]
```

```python
### Introducing DataLoader

from torch.utils.data import DataLoader

batch_size = 5 

train_dl = DataLoader(train_ds, batch_size, shuffle=True)

```

```python
model = nn.Linear(3, 2)
print(model.weight)

print(model.bias)
```

```python
inputs = torch.as_tensor(inputs, dtype=torch.float32)
```

```python
pred = model(inputs)

print(pred)
```
